```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
from sklearn.compose import ColumnTransformer
```

```python
X = df.drop(columns=["User_ID","Addiction_Level"])
y= df["Addiction_Level"]
```

```python
numeric_columns = X.select_dtypes(include="number").columns
```

```python
numeric_columns
```

```python
categorical_columns = X.select_dtypes(include=["string","category","object"]).columns
```

```python
categorical_columns
```

```python
numeric_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy="median")),
    ("scaler",StandardScaler())
])
```

```python
categorical_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy="most_frequent")),
    ("encoder",OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor = ColumnTransformer([
    ("numeric",numeric_pipeline,numeric_columns),
    ("categoric",categorical_pipeline,categorical_columns)
])
```

```python
model_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("regression",LogisticRegression(max_iter=200,class_weight="balanced",random_state=15))
])
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=15)
```

```python
model_pipeline.fit(X_train,y_train)
```

```python
y_pred = model_pipeline.predict(X_test)
```

```python
score = accuracy_score(y_pred,y_test)
print("Accuracy : ",score)
print(classification_report(y_pred,y_test))
print("Confusion Matrix : ",confusion_matrix(y_pred,y_test))
```

```python
from sklearn.model_selection import GridSearchCV,StratifiedKFold
```

```python
penalty = ["l1","l2","elasticnet"]
solver = ['liblinear', 'newton-cg', 'sag', 'saga' , 'lbfgs']
c_values = [100,10,1,0.1,0.01]
```

```python
params = dict(regression__penalty = penalty,regression__C=c_values,regression__solver=solver)
```

```python
cv= StratifiedKFold()
```

```python
grid = GridSearchCV(estimator=model_pipeline,param_grid=params,cv=cv,scoring="accuracy",n_jobs=-1)
```

```python
grid.fit(X_train,y_train)
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred,y_test)
print("Accuracy : ",score)
print(classification_report(y_pred,y_test))
print("Confusion Matrix : ",confusion_matrix(y_pred,y_test))
```
