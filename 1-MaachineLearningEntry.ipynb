```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
%matplotlib inline
```

```python
df = pd.read_csv('1-studyhours.csv')
```

```python
df
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model= LinearRegression() 
```

```python
X = df[["Study Hours"]]
y= df["Exam Score"]
```

```python
model.fit(X,y)
```

```python

yeni_veri = pd.DataFrame({
    "Study Hours": [5.5]
})
```

```python
tahmin = model.predict(yeni_veri)
print(tahmin)
```

```python
tahminler = model.predict(X)
plt.scatter(X["Study Hours"], y, label="Gerçek veriler")
plt.plot(
    X["Study Hours"],
    tahminler,
    label="Modelin tahmin doğrusu"
)

plt.xlabel("Çalışma Saati")
plt.ylabel("Sınav Puanı")
plt.title("Çalışma Saati ve Sınav Puanı")
plt.legend()
plt.show()
```
