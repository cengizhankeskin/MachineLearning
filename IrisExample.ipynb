## 4. Makine Öğrenmesi Araçlarının Hazırlanması / Preparing Machine-Learning Tools

**Türkçe:** Ön işleme, eğitim-test ayrımı, hiperparametre arama, sınıflandırma ve değerlendirme için gerekli scikit-learn bileşenleri içe aktarılır. Bu notebookta temel model Gaussian Naive Bayes'tir; Logistic Regression ve SVC karşılaştırma modeli olarak kullanılır.

**English:** The required scikit-learn components are imported for preprocessing, train-test splitting, hyperparameter search, classification, and evaluation. Gaussian Naive Bayes is the primary model, while Logistic Regression and SVC serve as comparison models.

```python
from sklearn.preprocessing import LabelEncoder,StandardScaler,OneHotEncoder
from sklearn.model_selection import GridSearchCV,train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix
```

## 5. Özellikler ve Hedef Değişken / Features and Target Variable

**Türkçe:** Girdi matrisi `X`, hedef sütun `Species` çıkarılarak oluşturulur. Böylece model yalnızca çiçeğin ölçümlerini açıklayıcı değişken olarak kullanır.

**English:** The input matrix `X` is created by removing the target column `Species`. This ensures that the model uses only flower measurements as explanatory variables.

```python
X=df.drop("Species",axis=1)
```

**Türkçe:** Tahmin edilmek istenen çiçek türü `y` hedef değişkenine atanır.

**English:** The flower species to be predicted is assigned to the target variable `y`.

```python
y=df["Species"]
```

## 6. Eğitim ve Test Kümelerinin Oluşturulması / Creating Training and Test Sets

**Türkçe:** Verinin %75'i model eğitimi, %25'i bağımsız değerlendirme için ayrılır. `random_state=15`, aynı bölünmenin tekrar üretilebilmesini sağlar. Mevcut kodda sınıf oranlarını koruyan `stratify=y` kullanılmamıştır.

**English:** Seventy-five percent of the data is reserved for model training and 25% for independent evaluation. `random_state=15` makes the split reproducible. The current code does not use `stratify=y` to preserve class proportions.

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=15)
```

## 7. Etiket Kodlama ve Ölçeklendirme Hazırlığı / Label Encoding and Scaling Setup

**Türkçe:** `Species` sütunundaki metin etiketleri DataFrame içinde sayısal değerlere dönüştürülür. Ancak `y` bu işlemden önce oluşturulduğu için mevcut modeller özgün metin etiketleriyle eğitilir; bu dönüşüm model girdisini değiştirmez.

**English:** Text labels in the DataFrame's `Species` column are converted to numeric values. However, because `y` was created before this operation, the current models are trained with the original text labels; this transformation does not change the model input.

```python
label = LabelEncoder()
df["Species"] = label.fit_transform(df["Species"])
```

**Türkçe:** Sayısal özellikleri sıfır ortalama ve birim standart sapmaya dönüştürmek için bir `StandardScaler` nesnesi hazırlanır.

**English:** A `StandardScaler` object is initialized to transform numerical features to zero mean and unit standard deviation.

```python
scaler = StandardScaler()
```

**Türkçe:** Ölçekleyici yalnızca eğitim verisine uyarlanır ve aynı dönüşüm test verisine uygulanır; bu yaklaşım veri sızıntısını önler. Bununla birlikte sonraki model hücreleri `X_train_scaled` ve `X_test_scaled` yerine ölçeklenmemiş verileri kullanmaktadır. Gaussian Naive Bayes için ölçekleme zorunlu değildir, ancak özellikle Logistic Regression ve SVC karşılaştırmalarında tutarlı kullanım önemlidir.

**English:** The scaler is fitted only on the training data, and the same transformation is applied to the test data, preventing data leakage. However, subsequent model cells use the unscaled data rather than `X_train_scaled` and `X_test_scaled`. Scaling is not mandatory for Gaussian Naive Bayes, but consistent use is especially important when comparing Logistic Regression and SVC.

```python
X_train_scaled  =scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## 8. Gaussian Naive Bayes Modeli / Gaussian Naive Bayes Model

**Türkçe:** Iris özellikleri sürekli sayısal değişkenler olduğu için Gaussian Naive Bayes seçilir. Model, her özelliğin her sınıf içinde yaklaşık normal dağıldığını ve özelliklerin sınıf koşuluyla bağımsız olduğunu varsayar.

**English:** Gaussian Naive Bayes is selected because the Iris features are continuous numerical variables. The model assumes that each feature is approximately normally distributed within each class and that features are conditionally independent given the class.

```python
naive = GaussianNB()
```

### Modelin Eğitilmesi / Training the Model

**Türkçe:** Model, eğitim örnekleri ve bunlara karşılık gelen tür etiketleri üzerinden sınıflara ait olasılık dağılımlarını öğrenir.

**English:** The model learns class-specific probability distributions from the training observations and their corresponding species labels.

```python
naive.fit(X_train,y_train)
```

### Test Kümesi Üzerinde Tahmin / Prediction on the Test Set

**Türkçe:** Eğitilen Naive Bayes modeli, daha önce görmediği test örneklerinin çiçek türlerini tahmin eder.

**English:** The trained Naive Bayes model predicts the flower species of previously unseen test observations.

```python
y_pred = naive.predict(X_test)
```

### Naive Bayes Performansının Değerlendirilmesi / Evaluating Naive Bayes Performance

**Türkçe:** Doğruluk genel başarı oranını; sınıflandırma raporu precision, recall ve F1-score değerlerini; karmaşıklık matrisi ise doğru ve hatalı tahminlerin sınıflara göre dağılımını gösterir. Not: scikit-learn metriklerinin geleneksel kullanımı `y_true, y_pred` sırasındadır; mevcut kod bu iki argümanı ters vermektedir. Accuracy ve confusion matrix'in temel sayımları değişmese de classification report içindeki precision ve recall yorumları yer değiştirebilir.

**English:** Accuracy gives the overall success rate; the classification report provides precision, recall, and F1-score; and the confusion matrix shows correct and incorrect predictions by class. Note: scikit-learn metrics conventionally expect arguments in the order `y_true, y_pred`; the current code supplies them in reverse. Although accuracy and the basic confusion-matrix counts remain related, precision and recall interpretations in the classification report may be interchanged.

```python
score = accuracy_score(y_pred,y_test)
print("Score : ",score)
print(classification_report(y_pred,y_test))
print("Confusion Matrix : \n ", confusion_matrix(y_pred,y_test))
```

## 9. Logistic Regression ile Karşılaştırma / Comparison with Logistic Regression

**Türkçe:** Aynı eğitim-test bölünmesi üzerinde bir Logistic Regression modeli kurularak Naive Bayes için karşılaştırmalı bir referans elde edilir.

**English:** A Logistic Regression model is built on the same train-test split to provide a comparative benchmark for Naive Bayes.

```python
#lOGİSTİC REGRESSİON
```

### Logistic Regression Modelinin Eğitilmesi / Training the Logistic Regression Model

**Türkçe:** Çok sınıflı sınıflandırma modeli eğitim verisine uyarlanır. Mevcut akış ölçeklenmemiş özellikleri kullanmaktadır.

**English:** The multiclass classification model is fitted to the training data. The current workflow uses unscaled features.

```python
model_logistic = LogisticRegression()
model_logistic.fit(X_train,y_train)
```

### Logistic Regression Tahminleri / Logistic Regression Predictions

**Türkçe:** Eğitilen model test kümesindeki tür etiketlerini tahmin eder.

**English:** The trained model predicts species labels for the test set.

```python
y_pred2 = model_logistic.predict(X_test)
```

### Logistic Regression Sonuçları / Logistic Regression Results

**Türkçe:** Modelin doğruluk, sınıf bazlı metrikler ve karmaşıklık matrisi yazdırılır. Burada da gerçek ve tahmin edilen etiketlerin metrik fonksiyonlarına ters sırada verildiği dikkate alınmalıdır.

**English:** The model's accuracy, class-level metrics, and confusion matrix are printed. Here again, note that the true and predicted labels are passed to the metric functions in reverse order.

```python
score = accuracy_score(y_pred2,y_test)
print("Score : ",score)
print(classification_report(y_pred2,y_test))
print("Confusion Matrix : \n ", confusion_matrix(y_pred2,y_test))
```

## 10. SVC ve Hiperparametre Optimizasyonu / SVC and Hyperparameter Optimization

**Türkçe:** Support Vector Classifier için farklı `C` ve çekirdek seçenekleri çapraz doğrulama ile karşılaştırılır. Amaç, test verisine bakmadan en başarılı hiperparametre kombinasyonunu seçmektir.

**English:** Different `C` and kernel choices for the Support Vector Classifier are compared using cross-validation. The goal is to select the strongest hyperparameter combination without inspecting the test data.

```python
#SVC
```

**Türkçe:** Varsayılan ayarlarla bir SVC nesnesi görüntülenir; bu hücre herhangi bir modeli eğitmez veya sonraki değişkenlere atamaz.

**English:** An SVC object with default settings is displayed; this cell does not train a model or assign it to a variable used later.

```python
SVC()
```

### Arama Uzayının Tanımlanması / Defining the Search Space

**Türkçe:** Düzenlileştirme parametresi `C` için beş değer; karar sınırı biçimi için doğrusal ve RBF çekirdekleri denenmek üzere tanımlanır.

**English:** Five values for the regularization parameter `C` and both linear and RBF kernels are defined for evaluation.

```python
param_grid = {
    "C":[0.01,0.1,1,10,100],
    "kernel" : ["linear","rbf"]
}
```

### GridSearchCV Yapılandırması / Configuring GridSearchCV

**Türkçe:** Her hiperparametre kombinasyonu 5 katlı çapraz doğrulamayla değerlendirilir ve seçim ölçütü olarak doğruluk kullanılır. SVC özellikle RBF çekirdeğinde ölçek farklılıklarına duyarlıdır; mevcut arama ölçeklenmemiş veriler üzerinde çalışır.

**English:** Each hyperparameter combination is evaluated with 5-fold cross-validation, using accuracy as the selection metric. SVC—especially with the RBF kernel—is sensitive to feature scales; the current search operates on unscaled data.

```python
grid = GridSearchCV(estimator=SVC(),param_grid=param_grid,scoring="accuracy",cv=5)
```

### Hiperparametre Aramasının Çalıştırılması / Running the Hyperparameter Search

**Türkçe:** Grid search eğitim verisi üzerinde tüm adayları dener ve çapraz doğrulama doğruluğu en yüksek modeli saklar.

**English:** Grid search evaluates all candidates on the training data and retains the model with the highest cross-validation accuracy.

```python
grid.fit(X_train,y_train)
```

### En İyi Model ve Test Tahminleri / Best Model and Test Predictions

**Türkçe:** Seçilen `C` ve çekirdek değerleri yazdırılır; ardından en iyi SVC modeli test kümesi için tahmin üretir.

**English:** The selected `C` and kernel values are printed, after which the best SVC model generates predictions for the test set.

```python
print(grid.best_params_)
y_pred3 = grid.predict(X_test)
```

### SVC Performansının Değerlendirilmesi / Evaluating SVC Performance

**Türkçe:** Optimize edilmiş SVC'nin test performansı diğer modellerle aynı metrikler üzerinden raporlanır. Metrik fonksiyonlarında argüman sırasının yine `y_pred, y_test` olduğu göz önünde bulundurulmalıdır. Modelleri karşılaştırırken yalnızca accuracy değil, sınıf bazlı F1-score ve hata dağılımı da değerlendirilmelidir.

**English:** The optimized SVC's test performance is reported using the same metrics as the other models. Keep in mind that the metric functions again receive arguments in the order `y_pred, y_test`. When comparing models, consider not only accuracy but also class-level F1-scores and the distribution of errors.

```python
score = accuracy_score(y_pred3,y_test)
print("Score : ",score)
print(classification_report(y_pred3,y_test))
print("Confusion Matrix : \n ", confusion_matrix(y_pred3,y_test))
```
